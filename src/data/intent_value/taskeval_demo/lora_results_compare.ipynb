```python
with open("llama3_8b_lora_finetune_filtered.json", 'r') as f:
    lora_llama = json.load(f)
    
with open("llama3_8b_finetune_actual.json", 'r') as f:
    actual = json.load(f)
    
with open("llama3_8b_finetune_filtered.json", 'r') as f:
    llama = json.load(f)
    
countries_list = ['Austria', 'England', 'France', 'Germany', 'Italy', 'Russia', 'Turkey']
```

```python
len(lora_llama), len(actual), len(llama)
```

```python
def remove_duplicates_by_uuid(dict_list, uuid_key='env_uuid'):
    seen = set()
    unique_list = []
    
    for d in dict_list:
        uuid_value = d.get(uuid_key)
        if uuid_value and uuid_value not in seen:
            seen.add(uuid_value)
            unique_list.append(d)
    
    return unique_list
```

```python
actual = remove_duplicates_by_uuid(actual)
print(len(actual))
# llama = llama[:141]
# print(len(llama))
```

```python
def get_score_with_env(countries, prediction):
    scores = 0
    for c in countries:
        scores += prediction[c][c.upper()][1]
    score_dict = {}
    score_dict['env'] = prediction['env_uuid']
    score_dict['scores'] = scores / len(countries)
    return score_dict

def get_score_dict_from_dialogue(dialogue):
    score_list = []
    for a in dialogue:
        if a['intent_dialogue'] == "":
            continue
        keys = []
        for vkeys in a.keys():
            if vkeys in countries_list:
                keys.append(vkeys)
        score_list.append(get_score_with_env(keys, a))
    return score_list
```

```python
actual_score = get_score_dict_from_dialogue(actual)
llama_lora_score = get_score_dict_from_dialogue(lora_llama)
llama_score = get_score_dict_from_dialogue(llama)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 提取每组数据中的分数
data1 = [d['scores'] for d in actual_score]
data2 = [d['scores'] for d in llama_lora_score]
data3 = [d['scores'] for d in llama_score]

# 合并所有数据以确定分数范围
all_data = data1 + data2 + data3

# 设置分数区间的个数（例如10个区间）
num_bins = 10
bins = np.linspace(min(all_data), max(all_data), num_bins + 1)

# 分别计算各组数据在各区间内的频数
counts1, _ = np.histogram(data1, bins=bins)
counts2, _ = np.histogram(data2, bins=bins)
counts3, _ = np.histogram(data3, bins=bins)

# 为每个区间创建标签，例如 "0.00-0.10"
bin_labels = [f"{bins[i]:.2f}-{bins[i+1]:.2f}" for i in range(len(bins) - 1)]

# 使用整数作为x轴分组位置，每个分数段对应一个位置
x = np.arange(len(bin_labels))
width = 0.25  # 每个柱子的宽度

plt.figure(figsize=(12, 6))
# 使用浅色系的配色方案
plt.bar(x - width, counts1, width=width, label='Actual Score', color='#A7C7E7', edgecolor='black')
plt.bar(x, counts2, width=width, label='Llama Score With LoRA', color='#77DD77', edgecolor='black')
plt.bar(x + width, counts3, width=width, label='Llama Score Without LoRA', color='#FFB347', edgecolor='black')

plt.xlabel('Score Range', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Scores', fontsize=15)
plt.xticks(x, bin_labels, rotation=45, fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```
